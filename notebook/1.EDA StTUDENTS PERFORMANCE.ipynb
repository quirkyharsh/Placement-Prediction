DATA CLEANING

```python
df = pd.read_csv('data/stud.csv')
```

```python
df.head()
```

```python
df = df.replace({'Yes': 1, 'No': 0, 'Placed': 1, 'NotPlaced': 0})
```

```python
df.head()
```

```python
# Save changes to a new file
df.to_csv('data/stud_processed.csv', index=False)
```

# Data Visulization


```python
dff = pd.read_csv('data/stud_processed.csv')
```

```python
sns.set_theme(style="whitegrid")
```

```python
plt.figure(figsize=(80, 45))
```

1. Distribution of Placement Status (Count Plot)

```python
plt.subplot(2, 2, 1)
sns.countplot(data=df, x='PlacementStatus', palette='Set2')
plt.title('Distribution of Placement Status (0 = Not Placed, 1 = Placed)')
plt.xlabel('Placement Status')
plt.ylabel('Number of Students')
```

2. Impact of CGPA on Placement (Box Plot)

```python
plt.subplot(2, 2, 2)
sns.boxplot(data=df, x='PlacementStatus', y='CGPA', palette='Pastel1')
plt.title('CGPA Distribution by Placement Status')
plt.xlabel('Placement Status (0: No, 1: Yes)')
plt.ylabel('CGPA')
```

3. CGPA vs Aptitude Test Score (Scatter Plot colored by Placement)

```python
plt.subplot(2, 2, 3)
sns.scatterplot(data=df, x='CGPA', y='AptitudeTestScore', hue='PlacementStatus', palette='deep', s=100)
plt.title('CGPA vs Aptitude Score by Placement Status')
plt.xlabel('CGPA')
plt.ylabel('Aptitude Test Score')
```
